```python
import pandas as pd
from pathlib import Path
from google_play_scraper import reviews, Sort
```

## **Menyimpan daftar aplikasi yang akan di Scraping**

```python
# Daftar aplikasi
apps = {
    "Claude": "com.anthropic.claude",
    "ChatGPT": "com.openai.chatgpt",
    "Gemini": "com.google.android.apps.bard",
    "Grok": "ai.x.grok",
    "Deepseek": "com.deepseek.chat"
}
```

# **Scraping data & Menyimpannya**

```python
all_reviews = []

for app_name, app_id in apps.items():
    print(f"Scraping ulasan {app_name}")

    result, _ = reviews(
        app_id,
        lang="id",
        country="id",
        sort=Sort.NEWEST,
        count=10000,
        filter_score_with=None
    )

    df = pd.DataFrame(result)

    columns = ["reviewId", "content", "score", "at"]
    df = df[columns]

    all_reviews.append(df)

    print(f"{app_name}: {len(df)} ulasan berhasil diambil")
```

```python
# Gabungkan & simpan seluruh ulasan
df_combined = pd.concat(all_reviews, ignore_index=True)

df_combined.to_csv(
    "ulasan_ai.csv",
    index=False,
    encoding="utf-8-sig"
)

print(f"Total ulasan: {len(df_combined)}")
print("File tersimpan di ulasan_ai.csv")
print(df_combined.head())
```
